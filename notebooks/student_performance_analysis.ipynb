```python
random_sample = df.sample(
    n=500,
    random_state=42
)
```

```python
stratified_sample = (
    df.groupby("Motivation_Level", group_keys=False)
      .sample(frac=500 / len(df), random_state=42)
)
```

```python
def compare_distribution(data, column):
    return (
        data[column]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
    )

comparison = pd.DataFrame({
    "Original": compare_distribution(
        df, "Motivation_Level"
    ),
    "Random Sample": compare_distribution(
        random_sample, "Motivation_Level"
    ),
    "Stratified Sample": compare_distribution(
        stratified_sample, "Motivation_Level"
    )
})

display(comparison)
```
